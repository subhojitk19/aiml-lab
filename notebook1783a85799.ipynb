```python
Develop  a python program to implement a Rule-Based Expert System for Student 
Performance Evaluation.The system should take inputs such as 
attendance,marks,assignment performance,and internal assessment and use IF THEN 
rules to classify the student as Excellent,Good,Average,or Poor.  
```

```python
 
print("STUDENT PERFORMANCE EVALUATION SYSTEM") 
print("--------------------------------------") 
 
attendance = float(input("Enter attendance percentage: ")) 
marks = float(input("Enter marks percentage: ")) 
assignment = float(input("Enter assignment performance percentage: ")) 
internal = float(input("Enter internal assessment percentage: ")) 
 
 
if (attendance >= 90 and marks >= 85 and 
    assignment >= 85 and internal >= 85): 
    performance = "Excellent" 
 
elif (attendance >= 80 and marks >= 70 and 
      assignment >= 70 and internal >= 70): 
    performance = "Good" 
 
elif (attendance >= 70 and marks >= 50 and 
      assignment >= 50 and internal >= 50): 
    performance = "Average" 
 
else: 
    performance = "Poor" 
 
print("\nStudent Performance:", performance)
```

Describe a python to implement a rule-based expert system for medical diagnosis.The 
system should accept symptoms such as fever,cough,hedache,and sore throat and use 
IF THEN rule to identify a possible disease.

```python
print("MEDICAL DIAGNOSIS EXPERT SYSTEM")
print("--------------------------------")

fever = input("Do you have fever? (yes/no): ").lower()
cough = input("Do you have cough? (yes/no): ").lower()
headache = input("Do you have headache? (yes/no): ").lower()
sore_throat = input("Do you have sore throat? (yes/no): ").lower()

if fever == "yes" and cough == "yes" and headache == "yes" and sore_throat == "yes":
    disease = "Flu (Influenza)"

elif fever == "yes" and cough == "yes" and sore_throat == "yes":
    disease = "Common Cold / Respiratory Infection"

elif fever == "yes" and headache == "yes":
    disease = "Possible Viral Infection"

elif cough == "yes" and sore_throat == "yes":
    disease = "Possible Throat Infection"

else:
    disease = "No specific disease identified"

print("\nPossible Diagnosis:", disease)
print("Note: This is only a rule-based prediction, not a medical diagnosis.")
```
